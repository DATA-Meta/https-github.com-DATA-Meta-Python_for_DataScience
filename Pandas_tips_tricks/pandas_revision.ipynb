# 1-🐼 Pandas Mastery — Lesson 1: DataFrames & Selection
🔹 1. What is a DataFrame?
A DataFrame is a table with rows and columns — like an Excel sheet but programmable.

```python
import pandas as pd
# creating dataframe
df = pd.DataFrame({
    'name':['brayn', 'khan', 'ali', 'shoaib', 'john'],
    'runs':[ 22, 55, 67, 78, 65],
    'strike_rate':[ 80, 100, 20, 59, 45]
    
})
df.head(3)
```

# 2- Column Selection

```python
# Use the same DataFrame you created earlier.

# Exercise 2.1
# Select only the name column.

df['name']
```

```python
#Exercise 2.2
# Select name and strike_rate together.

df[['name','strike_rate']]
```

```python
#Exercise 3.1
# Filter players with runs > 50.
df[(df['runs']> 50)]
```

```python
# # Exercise 3.2
# Filter players with:runs > 50
# strike_rate > 80
# (Use AND)

# df[(df['strike_rate'] > 80) & (df['runs'] > 50)]

df[(df['strike_rate'] > 80) | (df['runs'] > 50)]


```
